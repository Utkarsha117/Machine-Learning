```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=2,verbose=0,tol=1e-3,max_iter=300,n_init=20)
```

```python
# 1. Check data types and missing values
print(df.info())

# 2. If 'Private' contains text ('Yes'/'No'), ensure it's dropped, 
# and select only numeric columns explicitly:
X = df.drop('Private', axis=1, errors='ignore')
X = X.select_dtypes(include=[np.number])

# 3. Drop any rows with NaN values just in case
X = X.dropna()

# 4. Fit the KMeans model
kmeans.fit(X)
```

```python
def converter(cluster):
    if cluster=='Yes':
        return 1
    else:
        return 0
```

```python
from sklearn.metrics import confusion_matrix,classification_report
print(confusion_matrix(df1['Cluster'],kmeans.labels_))
print(classification_report(df1['Cluster'],kmeans.labels_))
```
