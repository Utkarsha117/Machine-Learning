```python
import matplotlib.pyplot as plt

# 1. Count plot: Survived by Sex
sns.countplot(data=titanic, x='Survived', palette='autumn', hue='Sex')
plt.show()

# 2. Count plot: Survived by Pclass
sns.countplot(data=titanic, x='Survived', palette='viridis', hue='Pclass')
plt.show()

# 3. Age Histogram
titanic['Age'].plot(kind='hist', bins=30, color='green')
plt.title('Age Distribution')
plt.show()

# 4. Scatter plot: Age vs Fare by Pclass
sns.scatterplot(data=titanic, x='Age', y='Fare', hue='Pclass')
plt.show()

# 5. Scatter plot: Age vs Fare by Sex
sns.scatterplot(data=titanic, x='Age', y='Fare', hue='Sex')
plt.show()
```

```python
# Calculate correlation matrix for numeric columns only
correlation = titanic.corr(numeric_only=True)
print(correlation)

# View correlation specifically for Survived
print("\nCorrelation with Survived:")
print(correlation['Survived'])

# Visualize correlation matrix heatmap
plt.figure(figsize=(12, 7))
sns.heatmap(correlation, annot=True, cmap='crest')
plt.show()
```
